```python
import requests
from rich.pretty import pprint
r = requests.get("https://www.iea.org/policies")
```

```python
from bs4 import BeautifulSoup
soup = BeautifulSoup(r.content)
links = soup.select("a.m-policy-listing-item__link")
links
```

```python
import pandas as pd
import numpy as np
data= []

for link in links:
    r = requests.get("https://iea.org" + link["href"])
    link_soup = BeautifulSoup(r.content)

    def get_items_by_label(label): # if not empty, collect all
        span = link_soup.find("span", class_="m-policy-content-list__title", string=label)
        return str([a.text.strip() for a in span.find_parent().find_all("a")]).replace("[","").replace("]","") if span and span.find_parent() else "NA"

    if len(link_soup.select("div.m-block p")) == 1: 
        data.append({"text": link_soup.select("div.m-block")[0].text.replace("\n",""), 
        "country": link_soup.select("span.o-policy-aside-item__value")[0].text, 
        "year": link_soup.select("span.o-policy-aside-item__value")[1].text, 
        "status": link_soup.select("span.o-policy-aside-item__value")[2].text,
        "jurisdiction": link_soup.select("span.o-policy-aside-item__value")[3].text, 
        "link": "https://iea.org" + link["href"],
        "topics":  get_items_by_label("Topics"), 
        "sectors": get_items_by_label("Sectors"),
        "policy types": get_items_by_label("Policy types"),
        "technologies": get_items_by_label("Technologies")
        })
    else:

        data.append({"text": link_soup.select("div.m-block p")[0].text, 
        "country": link_soup.select("span.o-policy-aside-item__value")[0].text, 
        "year": link_soup.select("span.o-policy-aside-item__value")[1].text, 
        "status": link_soup.select("span.o-policy-aside-item__value")[2].text,
        "jurisdiction": link_soup.select("span.o-policy-aside-item__value")[3].text, 
        "link": "https://iea.org" + link["href"], 
        "topics":  get_items_by_label("Topics"), 
        "sectors": get_items_by_label("Sectors"),
        "policy types": get_items_by_label("Policy types"),
        "technologies": get_items_by_label("Technologies")
        })

# extend to collect the whole database (not just the first page)
df = pd.DataFrame.from_dict(data)
df
```
